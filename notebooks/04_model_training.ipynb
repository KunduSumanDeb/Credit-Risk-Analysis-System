## 2. Load Module 3 Training Artifacts

The processed feature datasets and target splits generated during
Feature Engineering are loaded from the persisted project artifacts.

This allows Module 4 to run independently without depending on
variables remaining in the Module 3 notebook kernel.

```python
# =============================================================================
# Load Training Artifacts
# =============================================================================

from src.feature_engineering import load_pickle
from src.config import TRAINING_ARTIFACTS_PATH


training_artifacts = load_pickle(
    TRAINING_ARTIFACTS_PATH,
)


X_train = training_artifacts["X_train"]
X_validation = training_artifacts["X_validation"]
X_test = training_artifacts["X_test"]

y_train = training_artifacts["y_train"]
y_validation = training_artifacts["y_validation"]
y_test = training_artifacts["y_test"]


print("=" * 80)
print("Training Artifacts Loaded")
print("=" * 80)

print(f"X_train       : {X_train.shape}")
print(f"X_validation  : {X_validation.shape}")
print(f"X_test        : {X_test.shape}")

print(f"y_train       : {y_train.shape}")
print(f"y_validation  : {y_validation.shape}")
print(f"y_test        : {y_test.shape}")

print("=" * 80)
```

```python
# =============================================================================
# Validate Loaded Training Artifacts
# =============================================================================

required_artifacts = {
    "X_train": X_train,
    "X_validation": X_validation,
    "X_test": X_test,
    "y_train": y_train,
    "y_validation": y_validation,
    "y_test": y_test,
}


print("=" * 80)
print("Training Artifact Validation")
print("=" * 80)


for name, artifact in required_artifacts.items():

    if artifact is None:
        raise ValueError(
            f"Training artifact '{name}' cannot be None."
        )

    if len(artifact) == 0:
        raise ValueError(
            f"Training artifact '{name}' cannot be empty."
        )

    print(
        f"✅ {name:<15} "
        f"Type: {type(artifact).__name__:<12} "
        f"Rows: {len(artifact)}"
    )


if len(X_train) != len(y_train):
    raise ValueError(
        "X_train and y_train must contain the same number "
        "of observations."
    )


if len(X_validation) != len(y_validation):
    raise ValueError(
        "X_validation and y_validation must contain the same "
        "number of observations."
    )


if len(X_test) != len(y_test):
    raise ValueError(
        "X_test and y_test must contain the same number "
        "of observations."
    )


print("-" * 80)
print("✅ Feature-target lengths are consistent")
print("✅ All required training artifacts are valid")
print("=" * 80)
```

## 7. Probability Prediction

The `predict_probability()` function generates class probabilities
using a fitted classification model.

For the credit-risk problem, class 0 represents no default and
class 1 represents default.

The complete probability matrix is retained so that downstream
evaluation and risk-analysis components can use the appropriate
class probability.

```python
# =============================================================================
# Probability Prediction
# =============================================================================

from src.model_training import predict_probability
```

```python
# =============================================================================
# Generate Logistic Regression Probabilities
# =============================================================================

logistic_probabilities = predict_probability(
    trained_logistic_model,
    X_test,
)

print("=" * 80)
print("Logistic Regression Probability Prediction")
print("=" * 80)
print(f"Prediction Type : {type(logistic_probabilities).__name__}")
print(f"Prediction Shape: {logistic_probabilities.shape}")
print(f"Test Rows       : {len(X_test)}")
print(f"First 5 Values:")
print(logistic_probabilities[:5])
print("=" * 80)
```

```python
# =============================================================================
# Verify Probability Predictions
# =============================================================================

assert isinstance(
    logistic_probabilities,
    np.ndarray,
)

assert logistic_probabilities.shape == (
    len(X_test),
    2,
)

assert logistic_probabilities.ndim == 2

assert np.all(
    (logistic_probabilities >= 0)
    & (logistic_probabilities <= 1)
)

assert np.allclose(
    logistic_probabilities.sum(axis=1),
    1.0,
)

print("=" * 80)
print("Probability Prediction Verification")
print("=" * 80)
print("✅ Probabilities returned as NumPy array")
print("✅ Correct number of prediction rows")
print("✅ Two class probabilities returned")
print("✅ All probabilities are between 0 and 1")
print("✅ Class probabilities sum to 1")
print("=" * 80)
```

```python
# =============================================================================
# Inspect Default Probability
# =============================================================================

default_probabilities = logistic_probabilities[:, 1]

print("=" * 80)
print("Default Probability")
print("=" * 80)
print(f"Number of Applicants : {len(default_probabilities)}")
print(f"First 10 Probabilities:")
print(default_probabilities[:10])
print("=" * 80)
```

```python
# =============================================================================
# Test Probability Prediction With Unfitted Model
# =============================================================================

from sklearn.exceptions import NotFittedError


try:
    unfitted_model = build_logistic_regression()

    predict_probability(
        unfitted_model,
        X_test,
    )

except NotFittedError as error:
    print("=" * 80)
    print("Unfitted Model Probability Test")
    print("=" * 80)
    print("✅ Unfitted model correctly rejected")
    print(f"Error : {error}")
    print("=" * 80)
```

```python
# =============================================================================
# Final Model Integration Verification
# =============================================================================

models_and_outputs = {
    "Logistic Regression": (
        trained_logistic_model,
        logistic_predictions,
        logistic_probabilities,
    ),
    "Decision Tree": (
        trained_decision_tree,
        decision_tree_predictions,
        decision_tree_probabilities,
    ),
    "Random Forest": (
        trained_random_forest,
        random_forest_predictions,
        random_forest_probabilities,
    ),
}


print("=" * 80)
print("MODEL INTEGRATION VERIFICATION")
print("=" * 80)

for model_name, (
    model,
    predictions,
    probabilities,
) in models_and_outputs.items():

    assert len(predictions) == len(X_test)

    assert probabilities.shape == (
        len(X_test),
        2,
    )

    assert np.all(
        (probabilities >= 0)
        & (probabilities <= 1)
    )

    assert np.allclose(
        probabilities.sum(axis=1),
        1.0,
    )

    print(f"✅ {model_name}")
    print(f"   Model Type      : {type(model).__name__}")
    print(f"   Predictions     : {predictions.shape}")
    print(f"   Probabilities   : {probabilities.shape}")


print("=" * 80)
print("✅ ALL THREE MODELS PASSED INTEGRATION VERIFICATION")
print("=" * 80)
```
